## Why Do We Need Decisions?

Programs often need to make decisions based on information. For example:

- Has a simulation completed successfully?
- Is the temperature above a safe limit?
- Has a calculation converged?

Conditional statements allow Python to choose what action to perform.

## Comparison Operators

Python uses comparison operators to compare values.

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

Example:


````python
temperature = 26
print(temperature > 25)
````

> **Careful with `==` vs `=`:** a single `=` *assigns* a value to a variable (`temperature = 26`), while a double `==` *compares* two values and gives back `True` or `False` (`temperature == 26`). Mixing these up is one of the most common beginner mistakes — you'll see what happens if you do later in this lesson.

## A quick word on indentation

Before writing your first `if` statement, there's one Python rule that matters here more than it has so far: **indentation is not just for tidiness — it's how Python knows which lines belong together.**

Any line that should run "inside" an `if`, `for`, or `while` must be indented (normally by 4 spaces), and every line in that block must be indented by the *same* amount. Jupyter usually indents automatically for you after a line ending in `:`, but if you ever copy-paste code from elsewhere (including from a Word document!), it's worth checking the indentation has come through correctly — mismatched spaces and tabs are a common source of errors, which you'll see for yourself shortly.

## The if Statement

The simplest decision structure is the `if` statement.


````python
temperature = 28
if temperature > 25:
    print("Warning: High temperature")
````

If the condition is true, the message is displayed. Notice the **colon `:`** at the end of the `if` line, and the **indented** line underneath — both are required.

## Using else

Sometimes we want an alternative action.


````python
temperature = 20
if temperature > 25:
    print("High temperature")
else:
    print("Temperature is acceptable")
````

## Using elif

`elif` (short for "else if") allows several conditions to be checked in sequence.


````python
score = 75
if score >= 70:
    print("Distinction")
elif score >= 50:
    print("Pass")
else:
    print("Fail")
````

Python checks each condition in order and runs the first branch that matches — once one matches, it skips the rest. To see a different branch trigger, try a lower score:


````python
score = 40
if score >= 70:
    print("Distinction")
elif score >= 50:
    print("Pass")
else:
    print("Fail")
````

## Logical Operators

Logical operators allow multiple conditions to be combined.

| Operator | Meaning |
|---|---|
| `and` | Both conditions must be true |
| `or` | Either condition can be true |
| `not` | Reverses a condition |

Example:


````python
temperature = 24
pressure = 95

if temperature > 20 and pressure > 90:
    print("Conditions satisfied")
````

## Common Beginner Mistakes: Conditionals

It's worth deliberately breaking an `if` statement a few ways, so you recognise the errors when they happen to you by accident.

**Forgetting the colon:**


````python
temperature = 28
if temperature > 25
    print("Warning")
````

Python raises a `SyntaxError` and tells you it expected a `:`. Now try forgetting to indent the line underneath:

````python
temperature = 28
if temperature > 25:
print("Warning")
````

This time it's an `IndentationError` — Python expected an indented block and didn't find one.

**Using `=` instead of `==`:**


````python
temperature = 28
if temperature = 28:
    print("Warning")
````

Python catches this one too, and even suggests the fix in the error message itself: *"Maybe you meant '==' or ':=' instead of '='?"* This is a good example of an error message doing a lot of the debugging work for you — it's always worth reading the whole message, not just the error type.


## Repeating Tasks with for Loops

A `for` loop repeats instructions for every item in a collection.


````python
temperatures = [20.1, 21.5, 19.8]

for temp in temperatures:
    print(temp)
````

Notice the pattern is the same as `if`: a colon `:` at the end of the `for` line, and an indented block underneath that Python repeats once per item.

## Using range()

Sometimes we simply need to repeat an action a fixed number of times, without an existing list to loop over.


````python
for number in range(5):
    print(number)
````

Notice that `range(5)` produces the numbers **0, 1, 2, 3, 4** — five numbers starting from zero, not from one, and stopping *before* 5. This is the same zero-based counting you saw with list indexing in Lesson 3.

### range() with a start and stop value

`range()` can also take two numbers: a start and a stop. `range(1, 11)` counts from 1 up to (but not including) 11 — in other words, 1 to 10:


````python
for number in range(1, 11):
    print(number)
````

This two-argument form is exactly what you'll need for one of the exercises later — worth remembering that `range(start, stop)` always stops *before* reaching `stop`, the same "end excluded" rule you saw with slicing in Lesson 3.

## while Loops

A `while` loop continues until a condition becomes false.


````python
count = 1

while count <= 5:
    print(count)
    count = count + 1
````

Remember to update the variable controlling the loop (here, `count = count + 1`), otherwise the loop may never end.

> ⚠️ **Important safety note:** if you forget to update the loop variable, or write a condition that's always true, you create an **infinite loop** — the cell will run forever and Jupyter will look "stuck" (you'll see `[*]` next to the cell instead of a number). This is not a crash, and it's a very easy mistake to make. If it happens: use the ■ (stop) button in the toolbar, or the menu option **Kernel → Interrupt**, to stop it safely. Don't run the example below — it's shown only as text, not as a runnable cell, precisely because it never finishes:

```python
count = 1
while count <= 5:
    print(count)
    # forgot to update count here — this loop never stops!
```
