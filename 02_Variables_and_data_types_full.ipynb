## What is a Variable?

A variable is a named location used to store information.

Think of a variable as a labelled container that holds a value. For example:


````python
temperature = 25
````

In this example:

- `temperature` is the variable name.
- `25` is the value stored in the variable.

Notice that this line, on its own, produces no output — it just stores the value. We can then use the variable later:


````python
temperature = 25
print(temperature)
````

> **Why no output the first time?** Assigning a value (`temperature = 25`) is a different action from displaying a value (`print(temperature)`). This trips up a lot of beginners moving on from Lesson 1, where every line of code produced visible output. From now on, most lines *store or calculate* something quietly, and you'll use `print()` deliberately whenever you want to see a result.

## Naming Variables

Variable names should be meaningful and descriptive.

**Good examples:**


````python
temperature = 25
simulation_time = 120
sample_id = 105
````

**Poor examples:**


````python
x = 25
a = 120
z = 105
````

Although short names are sometimes useful, descriptive names make programs easier to understand — especially when you return to your own code weeks later, or share it with a colleague.

### Rules for Variable Names

Variable names:

- Must begin with a letter or underscore.
- Cannot begin with a number.
- Cannot contain spaces.
- Are case-sensitive.

**Valid:**


````python
sample_name = "A1"
temperature2 = 25
print(sample_name)
print(temperature2)
````

**Invalid — try it yourself.** The two lines below break the naming rules above. Run the cell and read the `SyntaxError` Python gives you:


````python
2temperature = 25
````

Now try the other invalid example:

````python
sample name = "A1"
````

Both fail for the reasons in the rules above: a name can't start with a digit, and it can't contain a space. Python is telling you exactly this in the error message — reading the error is often the fastest way to fix the problem.


## Common Data Types

Python provides several built-in data types. The four most common are:

| Data Type | Example | Purpose |
|---|---|---|
| Integer (`int`) | `10` | Whole numbers |
| Float (`float`) | `3.14` | Decimal numbers |
| String (`str`) | `"Research"` | Text |
| Boolean (`bool`) | `True` | True/False values |

### Integers

Integers represent whole numbers.


````python
cores = 64
jobs = 120
print(cores)
print(jobs)
````

Performing calculations:

````python
cores = 64
print(cores * 2)
````

### Floating-Point Numbers

Floats represent numbers containing decimal points.


````python
temperature = 21.5
runtime = 3.75
print(temperature)
print(runtime)
````

Calculations:

````python
runtime = 3.75
print(runtime * 2)
````

> **Reminder from Lesson 1:** dividing (or, as here, multiplying a float) often gives a result with a decimal point, like `7.5`. That's expected and correct — it isn't a mistake in your code.

### Strings

Strings contain text and are enclosed within quotation marks.


````python
project = "Climate Modelling"
username = "researcher01"
````

Displaying strings:

````python
project = "Climate Modelling"
print(project)
````

### Booleans

Boolean values represent either `True` or `False`.


````python
job_completed = True
error_detected = False
print(job_completed)
print(error_detected)
````

Booleans are commonly used when making decisions in programs (you'll meet this properly when you learn about `if` statements in a later lesson).

## Determining a Data Type

The `type()` function shows the type of a variable.


````python
temperature = 25
print(type(temperature))
````

Another example:

````python
temperature = 25.0
print(type(temperature))
````

Notice `<class 'int'>` versus `<class 'float'>` — this is simply Python's way of telling you the data type; you don't need to worry about the word "class" for now, just read it as "the type is int" or "the type is float".

## Type Conversion

Sometimes we need to convert data from one type to another.

### Convert Integer to Float


````python
value = 10
print(float(value))
````

### Convert Float to Integer

````python
value = 10.8
print(int(value))
````

> **Careful:** converting `10.8` to an integer gives `10`, not `11` — `int()` cuts off (truncates) the decimal part rather than rounding to the nearest whole number. If you actually want rounding, Python has a separate `round()` function, which is outside the scope of this lesson but worth remembering the name of for later.

### Convert Number to String


````python
value = 25
print(str(value))
````

## Arithmetic Operations

Python supports standard mathematical operations.

| Operator | Meaning |
|---|---|
| `+` | Addition |
| `-` | Subtraction |
| `*` | Multiplication |
| `/` | Division |
| `**` | Power |
| `%` | Remainder |

Example using the first four operators:


````python
a = 10
b = 3
print(a + b)
print(a - b)
print(a * b)
print(a / b)
````

The last line, `a / b`, gives `3.3333333333333335` — another example of division producing a float, as you'd now expect.

The table above also lists two operators the original example didn't show — `**` (power) and `%` (remainder). Here they are with the same `a` and `b`:


````python
a = 10
b = 3
print(a ** b)   # 10 to the power of 3
print(a % b)    # remainder after dividing 10 by 3
````

`a ** b` calculates \(10^3 = 1000\). `a % b` (the "modulo" operator) gives the remainder left over after division — `10 / 3` is `3` remainder `1`, so `a % b` is `1`. The remainder operator turns out to be surprisingly useful in HPC contexts, e.g. checking whether a job number is even or working out which of several nodes a task should be assigned to.
