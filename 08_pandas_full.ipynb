## Overview

Many research projects involve working with structured datasets, such as experimental measurements, simulation results or survey data. Whilst NumPy is excellent for numerical arrays, the **pandas** library provides more powerful tools for organising, analysing and manipulating tabular data.

pandas introduces the **DataFrame**, a flexible data structure that allows data to be organised into rows and columns, similar to a spreadsheet or database table. It also provides tools for reading data from files, selecting subsets of data, filtering results and calculating summary statistics.

In this lesson, you will learn how to create DataFrames, load data from CSV files, inspect datasets and perform basic data analysis.

## Why Use pandas?

Many research datasets are organised into tables. Examples include:

- Experimental measurements
- Climate observations
- Biological data
- Benchmark results
- Simulation summaries

Each row usually represents an observation, whilst each column stores a different variable.

Although NumPy arrays are excellent for numerical calculations, they do not provide meaningful column names or convenient methods for working with tabular data. pandas solves this problem by introducing labelled data structures.

## Importing pandas

Before using pandas, import the library.

## Creating a DataFrame

A DataFrame can be created from a Python dictionary.


```python
data = {
    "Sample": ["A", "B", "C"],
    "Temperature": [21.5, 22.1, 20.8],
    "Pressure": [101.3, 100.9, 101.8]
}

df = pd.DataFrame(data)
print(df)
```

Notice the numbers `0`, `1`, `2` down the left-hand side — this is the DataFrame's **index**, automatically added to label each row. You didn't create it yourself; pandas adds it by default.


## Setting Up Some Sample Data Files

The rest of this lesson reads from CSV files (`results.csv`, `experiment.csv`, `measurements.csv`). In real research work you'd already have these files — usually exported from an instrument, a simulation, or another program — but a brand-new notebook doesn't have them yet.

Run the cell below to create some sample CSV files to work with for the rest of the lesson (this uses `df.to_csv()`, which writes a DataFrame out to a file — the reverse of `read_csv()`, which you're about to meet):


```python
import pandas as pd

results_data = pd.DataFrame({
    "Sample": ["A", "B", "C"],
    "Temperature": [21.5, 22.1, 20.8],
    "Pressure": [101.3, 100.9, 101.8],
})
results_data.to_csv("results.csv", index=False)

experiment_data = pd.DataFrame({
    "Sample": ["E1", "E2", "E3", "E4", "E5", "E6", "E7", "E8"],
    "Temperature": [22.5, 26.1, 24.0, 27.8, 23.3, 25.9, 21.0, 28.4],
    "Runtime": [3.2, 4.1, 2.8, 5.0, 3.6, 4.4, 2.9, 5.3],
})
experiment_data.to_csv("experiment.csv", index=False)

measurements_data = pd.DataFrame({
    "SampleID": ["M1", "M2", "M3", "M4", "M5", "M6", "M7"],
    "Temperature": [19.8, 20.5, 21.2, 22.0, 20.1, 19.5, 21.8],
    "Pressure": [100.5, 101.0, 101.4, 100.8, 100.2, 101.1, 100.9],
})
measurements_data.to_csv("measurements.csv", index=False)

print("Sample files created: results.csv, experiment.csv, measurements.csv")
```

## Reading a CSV File

Many datasets are stored as comma-separated value (CSV) files.


```python
df = pd.read_csv("results.csv")
print(df)
```

`read_csv()` is one of the most commonly used pandas functions — it's often the very first line of a real analysis script.

## Inspecting a Dataset

Display the first few rows:


```python
print(df.head())
```

Display the last few rows:


```python
print(df.tail())
```

View information about the dataset:


```python
df.info()
```

> **A quirk worth knowing about:** unlike `head()` and `tail()`, `df.info()` already prints its summary directly — it doesn't hand back a value the way `mean()` or `head()` does. That means if you wrap it in `print()`, like `print(df.info())`, you'll see the summary *and then a stray `None`* underneath. Try it:


```python
print(df.info())
```

There's that `None` — it's `print()` dutifully displaying what `df.info()` returned (nothing), *after* `df.info()` had already printed its own summary as a side effect. This is exactly the same `return` vs. `print()` distinction from Lesson 5, just showing up somewhere new. The fix is simply not to wrap it: call `df.info()` on its own, as shown two cells above.

## Selecting Columns

Retrieve a single column:


```python
print(df["Temperature"])
```

Retrieve multiple columns (notice the **double square brackets** — one pair for "select from `df`", one pair for the list of column names):


```python
print(df[["Temperature", "Pressure"]])
```

> **Single vs. double brackets — why it matters:** `df["Temperature"]` (single column, single brackets) gives you back a pandas **Series** — a single labelled column. `df[["Temperature", "Pressure"]]` (a *list* of column names, hence the double brackets) gives you back a **DataFrame** — a table, even if it only ends up containing one column. You can see the difference yourself:


```python
print(type(df["Temperature"]))
print(type(df[["Temperature"]]))
```

This distinction rarely matters for just printing things, but it matters a lot once you start chaining further operations — some methods only exist on a DataFrame, others only on a Series.

## Selecting Rows

Use `iloc` to select rows by position.


```python
print(df.iloc[0])
```

Notice this looks quite different from the table view above — a single row is displayed "sideways", with each column name down the left and its value for that one row on the right. This is because `df.iloc[0]` also gives you back a Series (like a single column did), just built from one row instead of one column.

Retrieve several rows:


```python
print(df.iloc[0:3])
```

This slicing works the same way list slicing did in Lesson 3 and Lesson 4's `range()` — the end position isn't included, though here `0:3` happens to cover all 3 rows we have.

## Filtering Data

Suppose we only want temperatures greater than 21°C.


```python
warm = df[df["Temperature"] > 21]
print(warm)
```

This line does two things at once, which can look like unfamiliar magic the first time you see it. It's worth breaking apart:


```python
mask = df["Temperature"] > 21
print(mask)
```

`df["Temperature"] > 21` compares every value in the column to `21` and gives back a Series of `True`/`False` values — one per row. This is sometimes called a **boolean mask**. Then, `df[mask]` (or `df[df["Temperature"] > 21]` written in one line) uses that True/False pattern to keep only the rows where it's `True`:


```python
print(df[mask])
```

Same result as `warm` above — just built up in separate steps so you can see what's happening. Filtering like this allows researchers to focus on specific observations without writing a manual loop.

## Summary Statistics

Calculate the average temperature:


```python
print(df["Temperature"].mean())
```

Other useful functions include:

```python
print(df["Temperature"].min())
print(df["Temperature"].max())
print(df["Temperature"].std())
print(df["Temperature"].count())
```

These should look familiar — they're the same statistics NumPy gave you in Lesson 7 (`np.mean()`, `np.min()`, and so on), just called as methods on a column instead of functions applied to an array.
