## Why Work with Files?

Most scientific programs produce output files containing results, measurements or diagnostic information. Examples include:

- Experimental measurements
- Simulation outputs
- Benchmark results
- Log files
- Configuration files

Rather than entering data manually, Python can read these files automatically. Similarly, Python can write processed results to new files for later analysis.


## Setting Up Some Sample Files

The examples below read from files like `results.txt` and `measurements.txt` — but a brand-new notebook doesn't have those files yet! If you tried to open one before it existed, Python would give you an error (you'll actually see this error on purpose later in the lesson).

So, before going any further, run the cell below to create a few sample files to work with throughout this lesson. It uses `open()` in **write mode**, which you'll learn about properly in the "Writing to a File" section further down — for now, just run it so the rest of the notebook has something to read.


````python
with open("results.txt", "w") as file:
    file.write("Simulation completed successfully.")

with open("measurements.txt", "w") as file:
    file.write("12\n15\n18\n20\n")

with open("jobs.txt", "w") as file:
    file.write("Job 1: Complete\nJob 2: Complete\nJob 3: Failed\n")

print("Sample files created: results.txt, measurements.txt, jobs.txt")
````

## Opening a File

The `open()` function is used to open a file.


````python
file = open("results.txt", "r")
````

The `"r"` indicates that the file is opened for reading.

Common file modes include:

| Mode | Description |
|---|---|
| `"r"` | Read a file |
| `"w"` | Write a new file (overwrites existing content) |
| `"a"` | Append data to an existing file |

Don't forget to close the file when you're done with it — more on why shortly:


````python
file.close()
````

## Reading a File

To read the entire contents of a file:


````python
file = open("results.txt", "r")
contents = file.read()
print(contents)
file.close()
````

The `close()` function releases the file when it is no longer required. This matters more than it might seem: forgetting to close files can, over the course of a long program, leave many files "open" at once — on some systems this can eventually cause errors, and it's simply good practice to tidy up after yourself. You'll see a much safer way to handle this shortly (the `with` statement).

## Reading One Line at a Time

Often it is more efficient to process a file line by line — especially for very large files, where reading the whole thing into memory at once might not even be possible.


````python
file = open("measurements.txt", "r")

for line in file:
    print(line)

file.close()
````

> **Something looks odd here — extra blank lines!** This is a classic Python file-reading gotcha, not a mistake in your code. Each `line` read from the file already ends with an invisible newline character (`\n`) — and `print()` *also* adds its own newline after whatever you give it. So you end up with two newlines per line: one from the file, one from `print()`. This is worth understanding rather than just working around, because it explains a lot of confusing spacing bugs later on.
>
> The fix is to strip the extra newline off the end of each line before printing it, using `.strip()`:


````python
file = open("measurements.txt", "r")

for line in file:
    print(line.strip())

file.close()
````

That's much tidier. From here on, this notebook uses `.strip()` whenever printing lines read from a file — it's a good habit to adopt for your own code too.

# Writing to a File

To create a new file:


````python
file = open("summary.txt", "w")
file.write("Simulation completed successfully.")
file.close()
````

If the file already exists, its contents will be **replaced** — `"w"` mode doesn't ask for confirmation, so be careful you're not overwriting something you wanted to keep.

## Appending to a File

To add information without deleting existing content, use `"a"` (append) mode instead:


````python
file = open("summary.txt", "a")
file.write("\nNew results added.")
file.close()

# Let's check the result:
with open("summary.txt", "r") as file:
    print(file.read())
````

## Using the with Statement

The recommended approach is to use the `with` statement.


````python
with open("results.txt", "r") as file:
    contents = file.read()

print(contents)
````

Python automatically closes the file when the block finishes — even if an error happens partway through reading or writing. This reduces the likelihood of programming errors (like forgetting `file.close()`), which is why `with` is considered best practice and is what you should reach for by default from now on.

## Common Beginner Mistakes

**Trying to open a file that doesn't exist, in read mode:**


````python
with open("does_not_exist.txt", "r") as file:
    print(file.read())
````

Python raises a `FileNotFoundError` and even tells you exactly which filename it went looking for. This is usually either a typo in the filename, or the file genuinely hasn't been created yet (as in our very first cell in this lesson!).

**Trying to write to a file that's open in read mode:**


````python
file = open("results.txt", "r")
file.write("This should not work")
file.close()
````

This raises an `io.UnsupportedOperation: not writable` error — the file mode you opened with determines what you're allowed to do with it. If you need to both read and write, check the [Python file mode documentation](https://docs.python.org/3/library/functions.html#open) for combined modes — outside the scope of this lesson, but useful to know exists.

### Example 1: Reading a File



````python
with open("measurements.txt", "r") as file:
    print(file.read())
````

### Example 2: Writing Results

````python
average = 18.6

with open("average.txt", "w") as file:
    file.write(str(average))

# Check it worked:
with open("average.txt", "r") as file:
    print(file.read())
````

Notice `file.write()` needs a **string** — that's why `str(average)` converts the number first. Trying to write a number directly (`file.write(average)`) would raise a `TypeError`, similar to errors you've seen in earlier lessons when the wrong data type is used somewhere.

### Example 3: Processing a Log File


````python
with open("jobs.txt", "r") as file:
    for line in file:
        print(line.strip())
````
