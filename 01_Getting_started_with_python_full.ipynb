## Introduction

Python is one of the most widely used programming languages in research computing. It is popular because it is relatively easy to read, has a large scientific community, and provides access to many powerful libraries for data analysis, automation, visualisation and scientific computing.

In HPC, Python is often used to support research workflows. Researchers may use Python to prepare input files, process simulation outputs, analyse datasets, create plots, or automate repetitive tasks. Python is not always used to replace highly optimised code written in languages such as C, C++ or Fortran. Instead, it often acts as a flexible tool that helps researchers connect different parts of their workflow together.

| Research task | How Python can help |
|---|---|
| Preparing a simulation | Create or modify input files automatically. |
| Running repeated tasks | Automate commands that would otherwise be typed by hand. |
| Analysing output | Read data files and calculate useful results. |
| Sharing methods | Create scripts that can be saved, reused and shared. |

By completing this lesson, you will develop the confidence to write, save, and run your first Python programs and establish the foundations needed for the rest of the course.


## What is Python?

Python is a high-level programming language. A high-level language is designed to be read and written by people more easily than lower-level machine instructions. This makes Python a good language for beginners, but it is also powerful enough to be used in professional research and software development.

Python code often looks close to plain English. This is one of the reasons it is widely used in education and research. A readable language makes it easier to understand what a program is doing, return to old code, and share scripts with colleagues.

For example, the following line tells Python to display a message. Run the cell below:


````python
print("Hello, world!")
````

## Running Python

There are several ways to use Python. In this lesson, we will focus on two common approaches:

1. Using Python interactively.
2. Writing and running Python scripts.

Both approaches are useful. The best choice depends on what you are trying to do.

### Which Python to use

On most HPC systems there will be a default Python on your system. This is fine for now, but as you get more acquainted with Python, you may want to load a specific version that has all the required libraries you need. It is also good practice when submitting a job to a compute node to specify the Python version as well as any other required software. This is done with the standard `module load` command.

### Interactive Python

Interactive Python allows you to type one command at a time and immediately see the result. This is useful when you are learning, testing a small idea, or checking a quick calculation.

An interactive Python session on a terminal usually shows a prompt that looks like this:

```
>>>
```

You type your command after the prompt, e.g. `>>> print("Hello, world!")`.

**In this notebook, each code cell behaves like a single interactive command** — you run it and immediately see the result underneath, just like the `>>>` prompt on a terminal.

### When to use interactive Python

Interactive Python is useful when you want to:

- Try out a single command.
- Test a small calculation.
- Explore how a function works.
- Learn a new Python idea without creating a file.

However, interactive commands are not normally saved as a reusable program. For larger tasks or for work that needs to be repeated, it is better to write a script.

### Python scripts

A Python script is a text file containing Python code. The file usually has a name ending in `.py`, e.g. `hello.py`.

A script allows you to save your commands, edit them later, and run them again. This is very important in research computing because reproducibility matters. If you save your analysis as a script, you or another researcher can run the same steps again in the future.

You would run a script from the command line using:

```
python hello.py
```

**A Jupyter notebook sits between these two approaches**: like an interactive session, you see results immediately after each cell; like a script, the whole notebook can be saved, re-run from top to bottom, and shared with others.

| Approach | Best used for | Limitation |
|---|---|---|
| Interactive Python | Testing ideas and trying small commands. | Commands are not usually saved as a reusable program. |
| Python script | Saving, sharing, and repeating a program. | You need to create and manage a file. |
| Jupyter notebook | Testing ideas *and* keeping a saved, shareable record. | Cells can be run out of order, which can cause confusion if you're not careful (see note below). |

> **A note on running order:** In a notebook, cells don't have to run top-to-bottom — you could run cell 5 before cell 2. This is powerful for experimenting, but it can also cause confusing bugs if a later cell depends on an earlier one that hasn't been run yet (or has changed since). A good habit: periodically use the menu option **"Restart Kernel and Run All"** to check your notebook works cleanly from top to bottom — this is the notebook equivalent of running a script and is important for reproducibility.


````python
print("Hello, world!")
````

The `print()` function displays information on the screen. It is one of the first Python functions that most learners use because it provides immediate feedback.

You can use `print()` to display text, numbers, and simple calculations.

## Comments

Comments are notes written inside code. Python ignores comments when the program runs, but comments are useful for humans reading the code.

In Python, a comment begins with the `#` symbol.


````python
# Calculate the number of hours
print(48 / 2)
````

Python ignores the first line because it is a comment. It runs the second line because it is Python code.

### Why comments are useful

Comments help explain why a piece of code exists or what a section of code is intended to do. This is especially important in research computing because code may be reused, shared, reviewed, or returned to months later.

Good comments can help:

- You understand your own code when you return to it later.
- Colleagues understand your method.
- Future users reproduce your workflow.
- Reviewers or collaborators check your assumptions.

### Good and poor comments

A useful comment explains the purpose of the code:

```python
# Convert the simulation runtime from hours to days
print(72 / 24)
```

A less useful comment simply repeats the code without adding meaning:

```python
# Print 72 divided by 24
print(72 / 24)
```

As your programs become more complex, aim to write comments that explain *why* something is being done, not just *what* each line says.
