## Overview

Python lists are suitable for storing small collections of data, but scientific computing often involves processing thousands or even millions of numerical values. For these tasks, Python provides the **NumPy** library, which offers fast, memory-efficient arrays and a wide range of mathematical functions.

NumPy is the foundation of many scientific Python libraries and is widely used in research computing, data science and High Performance Computing. It enables researchers to perform calculations efficiently without writing explicit loops for every operation.

In this lesson, you will learn how to create NumPy arrays, perform mathematical operations, generate numerical data and calculate basic statistics.

## Why Use NumPy?

Scientific applications frequently work with very large numerical datasets. Examples include:

- Experimental measurements
- Simulation outputs
- Image data
- Time-series observations
- Machine learning datasets

Although Python lists can store numerical values, they are not optimised for large-scale numerical calculations. NumPy provides specialised arrays that are faster, more memory efficient and easier to manipulate.

## Creating Arrays

A NumPy array is created using the `array()` function.


````python
temperatures = np.array([20.1, 21.5, 19.8, 22.0])
print(temperatures)
````

> **A formatting quirk to expect:** you'll notice the output is `[20.1 21.5 19.8 22. ]` — no commas between values (unlike a Python list), and `22.0` is shown as `22.` with a trailing space rather than `22.0`. This is just NumPy's own way of neatly aligning decimal points across all the values in an array; it isn't an error and the underlying value is still exactly `22.0`. NumPy's display formatting takes a little getting used to — don't worry if it looks slightly different from what you might expect.

## Array Arithmetic

One of the major advantages of NumPy is that operations are applied to **every element automatically**.


````python
temperatures = np.array([20, 21, 22])
print(temperatures + 2)
````

Similarly:

````python
print(temperatures * 3)
````

No loops are required — compare this to Lesson 4, where you'd have needed a `for` loop and a new list to do the same thing to every item.

> **Why not just use a normal Python list here?** This is a good moment to see exactly what NumPy is giving you. Try the same operations on a plain Python list instead of a NumPy array:


````python
plain_list = [20, 21, 22]
print(plain_list * 3)
````

That's very different from what you might expect! Multiplying a Python list by `3` doesn't multiply each number — it **repeats the whole list three times**. And adding a number directly is even less forgiving:


````python
plain_list = [20, 21, 22]
print(plain_list + 2)
````

This raises a `TypeError` — Python lists simply don't support "add this number to every item" as a built-in operation. This is exactly the gap NumPy arrays fill: `array + 2` and `array * 3` do the sensible, element-by-element thing you'd want for numerical data, which is a big part of why NumPy exists.

## Creating Numerical Sequences

NumPy provides functions to generate arrays automatically.

### Using arange()


````python
numbers = np.arange(5)
print(numbers)
````

This works just like the `range()` you met in Lesson 4 (start at 0, stop *before* 5), but `arange()` gives you back a NumPy array instead of a plain sequence of numbers, ready for the kind of array arithmetic you just saw above.

### Using linspace()


````python
values = np.linspace(0, 10, 6)
print(values)
````

`linspace(0, 10, 6)` generates **6 evenly spaced values** between `0` and `10` (inclusive of both ends) — here, that's `0, 2, 4, 6, 8, 10`. This is different from `arange()`, which steps by a fixed amount and may or may not land exactly on the end value; `linspace()` instead lets you say exactly how many points you want, and it works out the spacing. It's commonly used when plotting graphs or generating sample points.

## Two-Dimensional Arrays

NumPy also supports matrices.


````python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(matrix)
````

## Array Shape

The shape of an array describes its dimensions.


````python
print(matrix.shape)
````

This array contains:

- 2 rows
- 3 columns

`shape` always reports `(rows, columns)` for a two-dimensional array — a useful thing to check whenever you're not sure how a dataset you've loaded is arranged.

### Bonus: indexing a 2D array

You can access individual elements or whole rows using square brackets, similar to list indexing from Lesson 3, but with two numbers separated by a comma: `array[row, column]`.


````python
print(matrix[0, 1])   # row 0, column 1
print(matrix[1])      # the whole of row 1
````

## Basic Statistics

NumPy includes many useful mathematical functions.


````python
data = np.array([12, 15, 18, 20])
print(np.mean(data))
````

Other common functions include:

````python
print(np.min(data))
print(np.max(data))
print(np.sum(data))
print(np.std(data))
````

`np.std()` calculates the **standard deviation** — a measure of how spread out the values are. You don't need to know the formula behind it for this lesson; just know it's a single function call away when you need it, rather than something you'd have to write yourself with a loop.
