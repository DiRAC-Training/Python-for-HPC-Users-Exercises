## Overview

As Python programs become larger and more complex, it is important to organise code into smaller, reusable sections. Functions allow us to group together related instructions, making programs easier to understand, test and maintain.

In research computing, functions are used to perform repeated calculations, process datasets, analyse simulation results and automate common tasks. Rather than writing the same code multiple times, a function can be written once and used whenever it is needed.

In this lesson, you will learn how to create and use functions, pass information into functions using parameters, return results to the main program, and write clear documentation to explain what your functions do.

## Why Use Functions?

Imagine you need to calculate the average temperature of several datasets. Without functions, you may repeat the same code many times. Instead, you can write the calculation once and reuse it whenever required.

Functions make programs:

- Easier to read.
- Easier to test.
- Easier to maintain.
- Less likely to contain errors.

## Creating a Function

Functions are created using the `def` keyword.


````python
def greet():
    print("Welcome to Python!")
````

This creates a function but does not execute it — running the cell above produces no output, because you've only *defined* the recipe, not followed it yet.

To run the function, you **call** it by writing its name followed by brackets:


````python
greet()
````

> **A very common mix-up:** writing `greet` (without the brackets) doesn't run the function — it just refers to the function itself. Try it:


````python
print(greet)
````

You'll see something like `<function greet at 0x...>` — Python's way of showing you "this is a function object", not its output. (The exact numbers after `0x` will differ every time you run it — that's just a memory address, and it isn't something you need to worry about.) The brackets `()` are what actually tells Python "run this now" — always double-check you haven't forgotten them.

## Parameters

Functions can receive information called **parameters**.


````python
def greet_person(name):
    print("Hello", name)

greet_person("Alice")
````

Parameters make functions more flexible because they can work with different values each time they are called:


````python
greet_person("Bob")
greet_person("Dr Chen")
````

## Returning Values

Many functions perform calculations and **return** a result.


````python
def square(number):
    return number * number

result = square(5)
print(result)
````

The `return` statement sends a value back to the part of the program that called the function — that's what makes it possible to store the result in a variable (`result`) and use it afterwards.

> **`return` vs `print()` — a crucial difference.** These look similar at first, but they do very different things. `print()` just *displays* something on screen; `return` actually *hands the value back* so the rest of your program can use it. See what happens if a function only prints instead of returning:


````python
def add_no_return(a, b):
    print(a + b)

outcome = add_no_return(3, 4)
print(outcome)
````

Notice `7` gets printed (from inside the function), but then `outcome` turns out to be `None` — Python's way of saying "nothing was returned". The function displayed an answer, but never actually gave one back. This is one of the most common bugs when people start writing functions: forgetting `return` and being surprised later that their variable is empty. If you want to both *see* a result immediately **and** *use* it later, you generally want `return`, and then `print()` the returned value separately (as in the `square()` example above).

## Multiple Parameters

Functions may accept several parameters.


````python
def add(a, b):
    return a + b

print(add(3, 7))
````

### Bonus: keyword arguments

So far we've relied on *position* — `add(3, 7)` means `a=3` and `b=7`, in that order. You can also name the parameters explicitly when calling a function, which can make the call easier to read, especially once a function has several parameters:


````python
print(add(a=3, b=7))
print(add(b=7, a=3))  # order no longer matters when you name them
````

**Try it yourself:** what happens if you call a function without giving it enough information? Run the cell below:


````python
add(3)
````

Python raises a `TypeError` telling you it's "missing 1 required positional argument: 'b'". The message even names the missing parameter — another example of an error message doing most of the diagnosis for you.

## Local and Global Variables

Variables created inside a function are called **local variables**. They exist only while the function is running.


````python
def example():
    value = 10
    print(value)

example()
````

Now try using that same variable name **outside** the function:


````python
print(value)
````

This produces a `NameError` — `value` only exists *inside* `example()`, for the brief time the function is running. As soon as the function finishes, that local variable is gone.

Variables created **outside** functions (like `temperature` or `runtimes` in earlier lessons) are called **global variables**, and are generally available throughout the program.

Whenever possible, functions should rely on **parameters** rather than global variables — it makes code easier to understand and reuse, because the function doesn't secretly depend on something defined elsewhere.

## Documentation Strings

Good programmers explain what their functions do. Python uses **docstrings** for this purpose — a short piece of text as the very first line inside the function, wrapped in triple quotes.


````python
def area(radius):
    """Calculate the area of a circle."""
    return 3.14159 * radius ** 2

print(area(2))
````

Docstrings make code easier for others (and future-you) to understand and maintain. They also show up automatically if someone asks Python for help about your function:


````python
help(area)
````
