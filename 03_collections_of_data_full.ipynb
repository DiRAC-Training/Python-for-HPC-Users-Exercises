## Why Use Collections?

Suppose we have five temperature measurements:


````python
temp1 = 20.1
temp2 = 21.5
temp3 = 19.8
temp4 = 22.0
temp5 = 20.7
````

Whilst this works, it quickly becomes difficult to manage large numbers of values — imagine doing this for 500 measurements instead of 5.

A collection allows us to store all related values together:


````python
temperatures = [20.1, 21.5, 19.8, 22.0, 20.7]
print(temperatures)
````

Collections make code simpler, more organised and easier to maintain.

## Lists

Lists are the most commonly used collection type in Python.

Lists:

- Store multiple values.
- Maintain the order of items.
- Can be modified after creation.

### Creating a List


````python
temperatures = [20.1, 21.5, 19.8, 22.0]
````

### Displaying a List

````python
temperatures = [20.1, 21.5, 19.8, 22.0]
print(temperatures)
````

## Accessing List Elements

Each item in a list has a position called an **index**.

Python uses **zero-based indexing** — the first item is at index `0`, not `1`.


````python
temperatures = [20.1, 21.5, 19.8, 22.0]
print(temperatures[0])
````

Index positions for this list:

| Value | Index |
|---|---|
| 20.1 | 0 |
| 21.5 | 1 |
| 19.8 | 2 |
| 22.0 | 3 |

## Modifying Lists

Lists are **mutable**, meaning they can be changed after creation.

> **A note on running order:** the cell below starts by defining `temperatures` fresh, rather than assuming it still has its original values from earlier in the notebook. This matters because if you'd already modified `temperatures` in an earlier cell, re-running things out of order could give you a different starting point than the one shown here. Starting each new demonstration from a fresh, known list keeps the output predictable — a habit worth carrying into your own notebooks.


````python
temperatures = [20.1, 21.5, 19.8, 22.0]
temperatures[0] = 20.5
print(temperatures)
````

## Adding Items

Note that this next cell also starts from a fresh `temperatures` list (matching the original, unmodified values), so the output below matches what you'd see if you ran this cell on its own.


````python
temperatures = [20.1, 21.5, 19.8, 22.0]
temperatures.append(21.0)
print(temperatures)
````

## Tuples

Tuples are similar to lists but **cannot be modified** after creation.

### Creating a Tuple


````python
coordinates = (12.5, 7.2)
````

### Accessing Values

````python
coordinates = (12.5, 7.2)
print(coordinates[0])
````

### Why Use Tuples?

Tuples are useful when values should not change. Examples:

- Geographic coordinates.
- Fixed simulation parameters.
- Version numbers.

**Try it yourself:** because tuples are immutable, trying to change one raises an error. Run the cell below and read the message Python gives you:


````python
coordinates = (12.5, 7.2)
coordinates[0] = 5.0
````

Python raises a `TypeError` telling you a tuple "does not support item assignment" — this is Python's way of enforcing the rule that tuples can't be changed once created. If you need to update a value, you'd need a list instead.


## Dictionaries

Dictionaries store data as **key-value pairs**.

### Creating a Dictionary


````python
sample = {
    "id": 101,
    "temperature": 21.5,
    "status": "Complete"
}
````

## Accessing Dictionary Values

Instead of a numeric index, you look up a value by its key:


````python
sample = {
    "id": 101,
    "temperature": 21.5,
    "status": "Complete"
}
print(sample["temperature"])
````

### Adding New Values

````python
sample = {
    "id": 101,
    "temperature": 21.5,
    "status": "Complete"
}
sample["operator"] = "Dr Smith"
print(sample)
````

### Why Use Dictionaries?

Dictionaries are useful when data has meaningful labels. Examples:

- Sample information.
- Job metadata.
- Configuration settings.

**Try it yourself:** looking up a key that doesn't exist raises an error. Run the cell below and read the message:


````python
sample = {"id": 101}
print(sample["temperature"])
````

Python raises a `KeyError` naming the missing key. This is a very common error when working with dictionaries — it usually means either a typo in the key name, or the key genuinely hasn't been added yet.


## Sets

Sets store **unique** values.

### Creating a Set


````python
research_groups = {"Physics", "Chemistry", "Biology"}
````

### Duplicate Values

````python
groups = {"Physics", "Physics", "Chemistry"}
print(groups)
````

Duplicate values are automatically removed — the set only keeps one copy of `"Physics"`.

> **A note on order:** unlike lists, sets do **not** keep track of the order you added items in. If your output above shows the items in a different order than shown here (e.g. `{'Chemistry', 'Physics'}` instead of `{'Physics', 'Chemistry'}`), that is completely normal — it does not mean anything is wrong with your code. Never rely on the order of items in a set.

### Why Use Sets?

Sets are useful for:

- Removing duplicates.
- Identifying unique values.
- Comparing collections.


## Indexing

Indexing allows us to access individual items.


````python
jobs = ["job1", "job2", "job3"]
print(jobs[1])
````

### Bonus: negative indexing

Python also lets you count from the *end* of a list using negative numbers — `-1` is the last item, `-2` is the second-to-last, and so on. This is a small addition beyond the core lesson, but it's used constantly in real code (e.g. grabbing the most recent result from a growing list of outputs):


````python
jobs = ["job1", "job2", "job3"]
print(jobs[-1])
````

## Slicing

Slicing allows multiple items to be selected at once.


````python
jobs = ["job1", "job2", "job3", "job4"]
print(jobs[1:3])
````

The start position is **included**, whilst the end position is **excluded** — so `jobs[1:3]` gives you the items at index `1` and `2`, but stops before index `3`. This trips a lot of people up at first, so it's worth re-reading that sentence once more before moving on.

## Common Beginner Mistakes

You've already seen two errors above (`TypeError` from changing a tuple, and `KeyError` from a missing dictionary key). Here's a third one that's just as common — accessing an index that doesn't exist:


````python
jobs = ["job1", "job2", "job3"]
print(jobs[10])
````

Python raises an `IndexError` telling you the "list index out of range". This usually means either you've miscounted (remembering that indexing starts at `0`), or the list has fewer items than you expected.

| Error | Usually means... |
|---|---|
| `TypeError` (on a tuple) | You tried to change something that isn't allowed to change. |
| `KeyError` | The dictionary key you asked for doesn't exist (check spelling and whether it's been added). |
| `IndexError` | The index you asked for is beyond the end of the list (or before the start, if very negative). |
