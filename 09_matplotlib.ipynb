## Why Visualise Data?

Researchers often need to answer questions such as:

- How does a measurement change over time?
- Which simulation performed best?
- Are two variables related?
- Are there any unusual results?

Graphs help answer these questions quickly and effectively.

Visualisation is widely used in:

- Scientific publications
- Technical reports
- Performance analysis
- Simulation validation
- Data exploration

## Importing Matplotlib

Import the plotting module using the standard alias.


```python
import matplotlib.pyplot as plt
```

The alias `plt` is the convention used throughout the Python community, just as `np` was for NumPy and `pd` was for pandas.

## Creating a Simple Line Graph


```python
temperature = [20, 21, 22, 23]
time = [1, 2, 3, 4]

plt.plot(time, temperature)
plt.show()
```

This creates a simple line graph. Matplotlib plots `time` along the x-axis and `temperature` along the y-axis, in the order the two lists are given — `plt.plot(x_values, y_values)`.

**Try it yourself:** what happens if the two lists don't have the same number of values?


```python
plt.plot([1, 2, 3], [10, 20])
plt.show()
```

A `ValueError` — Matplotlib can't draw a point without both an x and a y value, so it insists the two lists match in length. This is a very easy mistake to make once you're pulling `time` and `temperature` from two separately-edited lists or columns.

## Adding Titles and Labels

Good graphs should always include descriptive labels. Building on the example above, in the **same cell**:


```python
temperature = [20, 21, 22, 23]
time = [1, 2, 3, 4]

plt.plot(time, temperature)
plt.title("Temperature During Experiment")
plt.xlabel("Time (hours)")
plt.ylabel("Temperature (°C)")
plt.show()
```

Clear labels make figures much easier to interpret — a graph with unlabelled axes is much harder for anyone else (including future-you) to trust or understand.

## Adding a Grid


```python
temperature = [20, 21, 22, 23]
time = [1, 2, 3, 4]

plt.plot(time, temperature)
plt.title("Temperature During Experiment")
plt.xlabel("Time (hours)")
plt.ylabel("Temperature (°C)")
plt.grid(True)
plt.show()
```

Grid lines improve readability, especially when someone needs to read approximate values off the graph rather than just see the overall trend.

## Scatter Plots

Scatter plots show relationships between two variables, rather than connecting them with a line — useful when you want to see whether points are correlated, rather than implying a continuous progression between them.


```python
temperature = [20, 21, 22, 23]
time = [1, 2, 3, 4]

plt.scatter(time, temperature)
plt.show()
```

Scatter plots are commonly used to identify trends and correlations.

## Bar Charts

Bar charts compare categories.


```python
systems = ["A", "B", "C"]
runtime = [12, 8, 15]

plt.bar(systems, runtime)
plt.show()
```

## Setting Up a Sample Data File

The next section plots data straight from a pandas DataFrame. As in Lesson 8, we'll create a small sample CSV file first so there's something to load — run the cell below:


```python
import pandas as pd

temperature_log = pd.DataFrame({
    "Time": [0, 1, 2, 3, 4, 5],
    "Temperature": [20.0, 20.5, 21.2, 21.8, 22.1, 22.6],
})
temperature_log.to_csv("temperature_log.csv", index=False)

print("Sample file created: temperature_log.csv")
```

## Plotting Data from pandas

Matplotlib integrates directly with pandas — you can pass a DataFrame column straight into `plt.plot()`, exactly as you would a list.


```python
df = pd.read_csv("temperature_log.csv")

plt.plot(df["Time"], df["Temperature"])
plt.title("Temperature Over Time")
plt.xlabel("Time (hours)")
plt.ylabel("Temperature (°C)")
plt.show()
```

## Adding a Legend

Legends explain what each line or series on a graph represents — essential as soon as you have more than one line on the same axes.


```python
time = [1, 2, 3, 4]
sensor_a = [20, 21, 22, 23]
sensor_b = [19, 20, 21, 22]

plt.plot(time, sensor_a, label="Sensor A")
plt.plot(time, sensor_b, label="Sensor B")
plt.title("Temperature Readings from Two Sensors")
plt.xlabel("Time (hours)")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.show()
```

Each `plt.plot()` call in this cell drew onto the *same* set of axes (because they're in the same cell, before `plt.show()`), and the `label=` argument on each is what `plt.legend()` then displays. With just one line, `plt.legend()` still works, but a legend showing a single entry is rarely worth the space it takes up — it earns its place once you're comparing two or more series, as here.

## Saving Figures

Figures can be saved for reports and publications.


```python
time = [1, 2, 3, 4]
temperature = [20, 21, 22, 23]

plt.plot(time, temperature)
plt.title("Temperature During Experiment")
plt.xlabel("Time (hours)")
plt.ylabel("Temperature (°C)")
plt.savefig("temperature.png")
plt.show()

import os
print("Saved temperature.png —", os.path.getsize("temperature.png"), "bytes")
```

Common formats include:

- PNG — good general-purpose format, widely supported.
- PDF — vector format, scales to any size without becoming blurry; often preferred for print/publication.
- SVG — also vector-based; good for web use and further editing.

> **A habit worth adopting: save *before* you show.** In this cell, `savefig()` is called before `show()`. On some setups (particularly interactive, non-notebook ones), calling `show()` first can clear or close the figure, leaving `savefig()` with nothing to save. It generally isn't an issue inside a Jupyter notebook specifically, but `savefig()` before `show()` is the safe order to use everywhere, including in plain `.py` scripts — so it's worth building the habit here too.
